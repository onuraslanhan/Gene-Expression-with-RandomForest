```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

# 1. Reading Data

```python
df_train = pd.read_csv("train.csv")
df_actual = pd.read_csv("actual.csv")
```

# 2. Cleaning Data

```python
for i in range(1, 38):
    df_train = df_train.drop(columns = [f"call.{i}"])
df_train = df_train.drop(columns = ["call", "Gene Accession Number", "Gene Description"])
```

# 3. Adjusting and Aggregating Data

```python
df_train = df_train.T
df_train = df_train.reset_index().rename(columns={"index": "patient"})
df_actual["patient"] = df_actual["patient"].astype(int)
df_train["Cancer"] = df_actual["cancer"]
```

# 4. Converting Data Type

```python
df_train = pd.get_dummies(df_train, columns = ["Cancer"], drop_first = True)
df_train.head()
```

# 5. Train / Test Split

```python
x = df_train.drop(columns = "Cancer_AML")
y = df_train["Cancer_AML"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 21)
```

# 6. Modelling

```python
x_train.columns = x_train.columns.astype(str)
x_test.columns = x_test.columns.astype(str)

rfc = RandomForestClassifier(max_depth=3, min_samples_leaf=2, random_state=42)
model = rfc.fit(x_train, y_train)
model.score(x_test,y_test)
```

# 7. Results

```python
print("Train score:", model.score(x_train, y_train))
print("Test score:", model.score(x_test, y_test))
```
